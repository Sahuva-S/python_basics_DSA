```python
class Employee:
    def __init__(self):
        self.name = "Rahul"                 #public
        self._salary= 23                       #protected
        self.__password= "999999"              #private


    def show_details(self):
      print(self.name)
      print(self._salary)
      print(self.__password)

emp = Employee()
print(emp.name)         #works
print(emp._salary)      #works, but discouraged
                              #print(emp.__password)  #error
emp.show_details()
```
